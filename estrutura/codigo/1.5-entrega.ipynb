DOCUMENTOS CRIADOS

```r
docs <- c(
  d1 = "recuperacao de informacao ordena documentos por relevancia",
  d2 = "o modelo de espaco vetorial representa documentos como vetores",
  d3 = "bm25 e um modelo probabilistico de ranqueamento de texto",
  d4 = "aprendizado estatistico fundamenta a recuperacao moderna",
  d5 = "o indice invertido acelera a busca em muitos documentos",
  d6 = "embeddings capturam a semantica de palavras e documentos",
  d7 = "a avaliacao mede a relevancia dos resultados da busca",
  d8 = "ciencia de dados combina estatistica e programacao"
)
```

1. BITS PRA ACHAR 1 DOCUMENTO ENTRE N

```r
N <- length(docs)
log2(N)
```

2. QUANTO UMA PISTA VALE

```r
tok <- function(x) unlist(strsplit(x, " "))
contem <- function(t) names(docs)[sapply(docs, function(d) t %in% tok(d))]
contem("busca")
log2(8 / 2)
```

```r
contem("modelo")
log2(8 / length(contem("modelo")))
```

3. A FÓRMULA DA AUTOINFORMAÇÃO EM NÚMEROS

```r
I <- function(p) -log2(p)
I(c(1, 1/2, 1/4, 1/8))
```

4. IDF EM BITS PRA TERMOS ESPECÍFICOS

```r
df <- c(modelo = 2, de = 5, recuperacao = 2)
round(log2(8 / df), 3)
```

```r
df2 <- c(modelo = 2, de = 5, recuperacao = 2, bm25 = 1)
round(log2(8 / df2), 3)
```

5. MONTANDO A TABELA DE BITS DE UMA CONSULTA INTEIRA

```r
tokens <- lapply(docs, tok)
vocab <- sort(unique(unlist(tokens)))
tf <- sapply(tokens, function(t) as.integer(table(factor(t, levels = vocab))))
rownames(tf) <- vocab
df <- rowSums(tf > 0)
idf_bits <- log2(8 / df)

termos_q <- c("modelo", "de", "recuperacao")
tab <- sapply(termos_q, function(t) tf[t, ] * idf_bits[t])
cbind(tab, total = rowSums(tab))
```

6. QUANDO OS BITS DE DUAS PISTAS SOMAM DE FORMA CORRETA

```r
intersect(contem("busca"), contem("documentos"))
log2(8/2) + log2(8/4)
```

```r
intersect(contem("busca"), contem("muitos"))
log2(8/2) + log2(8/1)
```

7. QUANDO A CONTA NÃO FECHA

```r
intersect(contem("recuperacao"), contem("relevancia"))
log2(8/2) + log2(8/2)
```

```r
intersect(contem("estatistico"), contem("aprendizado"))
log2(8/1) + log2(8/1)
```

8. O CASO EXTREMO DAS ROTAS (COLAPSO TOTAL)

```r
rotas <- c(
  r1 = "santos cubatao guaruja bertioga", r2 = "santos guaruja cubatao bertioga",
  r3 = "santos bertioga guaruja cubatao", r4 = "cubatao santos bertioga guaruja",
  r5 = "guaruja bertioga santos cubatao", r6 = "bertioga cubatao santos guaruja",
  r7 = "guaruja santos cubatao bertioga", r8 = "bertioga guaruja cubatao santos"
)
tokens <- lapply(rotas, function(x) unlist(strsplit(x, " ")))
vocab <- sort(unique(unlist(tokens)))
tdm <- sapply(tokens, function(t) as.integer(table(factor(t, levels = vocab))))
rownames(tdm) <- vocab
df <- rowSums(tdm > 0)
idf <- log(8 / df)
idf
```
